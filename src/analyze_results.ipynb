```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

data_gpt4o = [
    {'tags': ['AF_abcd_gpt-4o'], 'sum_totalCost': 1.000412499955},
    {'tags': ['RF_abcd_gpt-4o'], 'sum_totalCost': 0.998722499954},
    {'tags': ['NR_abcd_gpt-4o'], 'sum_totalCost': 0.738172499992},
]

data_gpt54 = [
    {'tags': ['RF_abcd_gpt-5.4'], 'sum_totalCost': 1.100382499964},
    {'tags': ['AF_abcd_gpt-5.4'], 'sum_totalCost': 1.091697499957},
    {'tags': ['NR_abcd_gpt-5.4'], 'sum_totalCost': 0.771687499962},
]

data_o3 = [
    {'tags': ['RF_abcd_o3'], 'sum_totalCost': 1.05550599993},
    {'tags': ['AF_abcd_o3'], 'sum_totalCost': 1.044015999935},
    {'tags': ['NR_abcd_o3'], 'sum_totalCost': 0.866641999943},
]

data = data_gpt4o + data_gpt54 + data_o3

df = pd.DataFrame(data)

# Extract variant + model
df["tag"] = df["tags"].apply(lambda x: x[0])
df["variant"] = df["tag"].str.split("_").str[0]
df["model"] = df["tag"].str.split("_").str[-1]

# Convert to cents
df["cost_cents"] = df["sum_totalCost"] * 100

df
```

```python
import numpy as np
import matplotlib.pyplot as plt

df["tag"] = df["tags"].apply(lambda x: x[0])
df["variant"] = df["tag"].str.split("_").str[0]
df["model"] = df["tag"].str.split("_").str[-1]
df["cost_cents"] = df["sum_totalCost"] * 100

# model on x-axis, variants as grouped bars
pivot_df = df.pivot(index="model", columns="variant", values="cost_cents")

# ensure consistent order
pivot_df = pivot_df.reindex(columns=["AF", "RF", "NR"])

variants = ["AF", "RF", "NR"]
models = pivot_df.index.tolist()

bar_width = 0.25
group_spacing = 0.5
x = np.arange(len(models)) * (len(variants) * bar_width + group_spacing)

colors = {
    "AF": "#2F238D",
    "RF": "#35D0A7",
    "NR": "#DE195BEA",
}

plt.figure(figsize=(8, 5))
plt.ylim(top=126)

for i, variant in enumerate(variants):
    values = pivot_df[variant].values
    positions = x + i * bar_width
    bars = plt.bar(
        positions,
        values,
        width=bar_width,
        label=variant,
        color=colors[variant]
    )
    plt.bar_label(bars, fmt="%.1f", padding=2)

plt.xticks(x + bar_width, models)
plt.ylabel("Cost (cents)")
plt.title("ABCD: Cost comparison by model (grouped by variant)")
plt.legend(loc="lower left", bbox_to_anchor=(0, 0))
plt.tight_layout()
plt.show()
```
